```python
from keras.datasets import imdb
from keras.preprocessing.text import Tokenizer
from keras.utils import pad_sequences
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
history = model.fit(X_train, y_train,epochs=5,validation_data=(X_test,y_test))
```

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

text = """I love learning machine learning because it helps computers
understand data and make intelligent predictions.
Machine learning is an important part of artificial intelligence.
Deep learning uses neural networks to learn complex patterns.
LSTM is a special type of recurrent neural network.
LSTM can remember important information from previous words.
Natural language processing helps computers understand human language.
I love learning deep learning and natural language processing.
Machine learning is used in healthcare education and business.
Artificial intelligence makes computers more intelligent.
Deep learning is useful for image recognition and text prediction.
LSTM is useful for next word prediction and sequence learning."""

# Convert words into numbers
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

# Get total number of unique words
total_words = len(tokenizer.word_index) + 1

# Generate sequences for next word prediction
#The loop creates:
#i	tokens[:i+1]	Sequence
#1	tokens[:2]	[1, 2]
#2	tokens[:3]	[1, 2, 3]
#3	tokens[:4]	[1, 2, 3, 4]
sequences = []

for line in text.split('\n'):
    tokens = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(tokens)):
        sequences.append(tokens[:i+1])

# Find maximum sequence length
max_len = max(len(seq) for seq in sequences)

# Make all sequences equal in length
sequences = pad_sequences(sequences, maxlen=max_len, padding='pre')

# X contains previous words and y contains the next word
X = sequences[:, :-1]
y = sequences[:, -1]

# Create LSTM model
model = Sequential()

# Convert word numbers into vectors of 10 values
model.add(Embedding(total_words, 10))

# Learn patterns from previous words
model.add(LSTM(50))

# Predict the next word from the vocabulary
model.add(Dense(total_words, activation='softmax'))

# Configure the model
model.compile(loss='sparse_categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])

# Train the model
model.fit(X, y, epochs=500)

print("Model Training Completed!")

# Starting sentence
seed_text = "I love learning"

# Predict the next 5 words
for i in range(5):

    # Convert sentence into numbers
    tokens = tokenizer.texts_to_sequences([seed_text])[0]

    # Apply padding
    tokens = pad_sequences([tokens], maxlen=max_len-1, padding='pre')

    # Predict the next word
    prediction = model.predict(tokens)

    # Select the word with the highest probability
    predicted_index = np.argmax(prediction)

    # Convert number back into a word
    next_word = tokenizer.index_word.get(predicted_index, '')

    # Add predicted word to the sentence
    seed_text += " " + next_word

print("Predicted Sentence:", seed_text)
```
